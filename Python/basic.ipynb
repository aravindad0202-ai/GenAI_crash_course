# Functions

```python
def add_value():
    return 'hi'

add_value()
```

```python
def add_value(a: int, b: int, c:int) -> int:
    """Add 2 value"""
    return a+b


add_value('adsf','adsf', 'asdf') # Positional data passing to an function, Order is important
add_value(b=2, a=1, c=7) # assigning value by initating the argument, Position is not important
add_value(3, b=5, c=3)
```

# *args, **kwargs

```python
def add_value(*args):
    # We can pass 'n' no of input values
    print(args[0]) # Index based retrievel

add_value(2,3,4,5)
```

```python
def add_value(**kwargs):
    for i , j in kwargs.items():
        print(i, j)


add_value(a=1,b=2)
```

```python
a = ['abcd.pdf', 'defg.pdf', 'asdf.txt']
pdf = []
for i in a:
    if i.endswith('.pdf'):
        pdf.append(i)

print(pdf)
```

```python
a = ['abcd.pdf', 'defg.pdf', 'asdf.txt']

pdf = list(filter(lambda a: a.endswith('.pdf'), a))
print(pdf)
```

```python
a = [1,2,3,4,5]

sdf = list(map(lambda x: x+2, a))
print(sdf)
```
